```python

# Step 37.3: Normalize ISBNs for comparison

import re

def normalize_isbn(isbn):
    if pd.isna(isbn):
        return None

    isbn = str(isbn).strip().upper()

    # Keep only digits and X
    isbn = re.sub(r'[^0-9X]', '', isbn)

    return isbn if isbn else None


# Create copies so original datasets remain unchanged
popularity_check = popularity_df.copy()
books_check = books.copy()

# Normalize ISBNs in both datasets
popularity_check["ISBN_normalized"] = (
    popularity_check["ISBN"].apply(normalize_isbn)
)

books_check["ISBN_normalized"] = (
    books_check["ISBN"].apply(normalize_isbn)
)

# Check normalized ISBN matches
book_isbn_set = set(
    books_check["ISBN_normalized"].dropna()
)

popularity_check["normalized_match"] = (
    popularity_check["ISBN_normalized"].isin(book_isbn_set)
)

print("Original matched books:", 149836)

print(
    "Matched after normalization:",
    popularity_check["normalized_match"].sum()
)

print(
    "Additional matches:",
    popularity_check["normalized_match"].sum() - 149836
)
```

```python

# Step 39.5: Iterative filtering

# Starting with explicit ratings
cf_ratings = explicit_ratings.copy()

# Set minimum thresholds
MIN_USER_RATINGS = 5
MIN_BOOK_RATINGS = 5

previous_size = -1

# Repeat until the dataset stops changing
while previous_size != len(cf_ratings):

    previous_size = len(cf_ratings)

    # Retain users with at least 5 ratings
    user_counts = cf_ratings.groupby("User-ID").size()

    active_users = user_counts[
        user_counts >= MIN_USER_RATINGS
    ].index

    cf_ratings = cf_ratings[
        cf_ratings["User-ID"].isin(active_users)
    ]

    # Retain books with at least 5 ratings
    book_counts = cf_ratings.groupby("ISBN").size()

    active_books = book_counts[
        book_counts >= MIN_BOOK_RATINGS
    ].index

    cf_ratings = cf_ratings[
        cf_ratings["ISBN"].isin(active_books)
    ]


# Final dataset statistics
print("Original explicit ratings:", len(explicit_ratings))

print("Filtered ratings:", len(cf_ratings))

print("Retained users:", cf_ratings["User-ID"].nunique())

print("Retained books:", cf_ratings["ISBN"].nunique())

print(
    "Percentage of ratings retained:",
    round(len(cf_ratings) / len(explicit_ratings) * 100, 2),
    "%"
)

```

```python

# Step 39.6: Create Sparse User-Book Matrix

import numpy as np
from scipy.sparse import csr_matrix

# Check for duplicate user-book rating pairs
duplicate_count = cf_ratings.duplicated(
    subset=["User-ID", "ISBN"]
).sum()

print("Duplicate user-book pairs:", duplicate_count)

# If duplicates exist, combine them using their average rating
matrix_data = cf_ratings.groupby(
    ["User-ID", "ISBN"],
    as_index=False
)["Book-Rating"].mean()

# Create mappings for users and books
user_ids = matrix_data["User-ID"].unique()
book_isbns = matrix_data["ISBN"].unique()

user_to_index = {
    user_id: index
    for index, user_id in enumerate(user_ids)
}

book_to_index = {
    isbn: index
    for index, isbn in enumerate(book_isbns)
}

# Convert user IDs and ISBNs into matrix positions
row_indices = matrix_data["User-ID"].map(
    user_to_index
).values

column_indices = matrix_data["ISBN"].map(
    book_to_index
).values

ratings_values = matrix_data["Book-Rating"].values

# Construct sparse matrix
user_book_matrix = csr_matrix(
    (
        ratings_values,
        (row_indices, column_indices)
    ),
    shape=(len(user_ids), len(book_isbns))
)

# Calculate matrix statistics
total_cells = (
    user_book_matrix.shape[0] *
    user_book_matrix.shape[1]
)

density = (
    user_book_matrix.nnz / total_cells
) * 100

print("\nUser-Book Matrix Information")

print("Matrix shape:", user_book_matrix.shape)

print("Stored ratings:", user_book_matrix.nnz)

print("Total possible entries:", total_cells)

print("Matrix density:", round(density, 4), "%")

print("Matrix type:", type(user_book_matrix))

```

```python

import numpy as np

# Convert matrices into NumPy arrays
similarity_matrix = np.asarray(train_user_similarity)

rating_matrix = train_user_book_matrix.toarray()

# Store users who have rated each book
book_raters = {}

for book_idx in test_eval["book_idx"].unique():
    book_raters[book_idx] = np.where(
        rating_matrix[:, book_idx] > 0
    )[0]

# Number of nearest neighbours
K = 20

predictions = []

# Generate predictions
for user_idx, book_idx in test_eval[
    ["user_idx", "book_idx"]
].itertuples(index=False, name=None):

    # Find users who rated this book
    raters = book_raters[book_idx]

    # Exclude the target user
    raters = raters[raters != user_idx]

    # Get similarity scores
    similarities = similarity_matrix[user_idx, raters]

    # Keep only users with positive similarity
    valid = similarities > 0
    raters = raters[valid]
    similarities = similarities[valid]

    if len(raters) == 0:
        predictions.append(np.nan)
        continue

    # Select the 20 most similar users
    top_k = np.argsort(similarities)[-K:]

    nearest_users = raters[top_k]
    nearest_similarities = similarities[top_k]

    # Get their ratings for this book
    neighbour_ratings = rating_matrix[
        nearest_users, book_idx
    ]

    # Calculate weighted average
    predicted_rating = np.sum(
        nearest_similarities * neighbour_ratings
    ) / np.sum(nearest_similarities)

    predictions.append(predicted_rating)

# Store predictions
test_eval["predicted_rating"] = predictions

print("Prediction generation completed!")

print("Total testing ratings:", len(test_eval))
print("Successful predictions:", test_eval["predicted_rating"].notna().sum())
print("Failed predictions:", test_eval["predicted_rating"].isna().sum())
```

```python
# Step 13: Create Content-Based Recommendation Function

def get_content_recommendations(book_title, top_n=10):

    # Find the book in the dataset (case-insensitive)
    matching_indices = content_books.index[
        content_books['Book-Title'].str.lower() ==
        book_title.lower()
    ].tolist()

    # Check whether the book exists
    if len(matching_indices) == 0:
        print("Book not found in the dataset!")
        return

    # Select the first matching book
    book_index = matching_indices[0]

    # Calculate similarity with all books
    similarity_scores = cosine_similarity(
        tfidf_matrix[book_index],
        tfidf_matrix
    ).flatten()

    # Exclude books having the same title
    for idx in matching_indices:
        similarity_scores[idx] = -1

    # Get indices of the most similar books
    similar_indices = similarity_scores.argsort()[::-1][:top_n]

    # Display recommendations
    recommendations = content_books.iloc[similar_indices][
        ['Book-Title', 'Book-Author', 'Publisher']
    ].copy()

    recommendations['Similarity Score'] = similarity_scores[similar_indices]

    return recommendations.reset_index(drop=True)


print("Content-Based Recommendation Function created successfully!")
```

```python

# Create Hybrid recommendations using existing models

hybrid_results = []

# Convert content candidates into a convenient lookup
content_groups = content_candidates_df.groupby("User-ID")

# Create ISBN-to-index mapping for Item-CF
isbn_to_item_index = book_mapping

# Prepare book information
book_details_lookup = books.drop_duplicates("ISBN")[
    ["ISBN", "Book-Title", "Book-Author", "Publisher"]
].copy()

# Give equal importance to both recommendation approaches
content_weight = 0.5
collaborative_weight = 0.5

for user_id, user_candidates in content_groups:

    user_idx = user_mapping.get(user_id)

    if user_idx is None:
        continue

    # Retrieve books already rated by this user
    user_ratings = train_user_book_matrix[user_idx].toarray().flatten()

    rated_indices = np.where(user_ratings > 0)[0]
    rated_books = set(rated_indices)

    # Convert candidate ISBNs into Item-CF indices
    candidate_indices = {}

    for _, row in user_candidates.iterrows():
        book_idx = isbn_to_item_index.get(row["ISBN"])

        if book_idx is not None and book_idx not in rated_books:
            candidate_indices[book_idx] = row["ISBN"]

    # Store collaborative scores for candidate books
    weighted_rating_sum = {}
    similarity_sum = {}

    # Find candidate books among the neighbours of previously rated books
    for rated_idx in rated_indices:

        neighbour_indices = item_indices_50[rated_idx]
        neighbour_similarities = item_similarities_50[rated_idx]

        for neighbour_idx, similarity in zip(
            neighbour_indices, neighbour_similarities
        ):

            neighbour_idx = int(neighbour_idx)

            if neighbour_idx not in candidate_indices:
                continue

            if similarity <= 0:
                continue

            rating = user_ratings[rated_idx]

            weighted_rating_sum[neighbour_idx] = (
                weighted_rating_sum.get(neighbour_idx, 0)
                + similarity * rating
            )

            similarity_sum[neighbour_idx] = (
                similarity_sum.get(neighbour_idx, 0)
                + similarity
            )

    # Calculate Hybrid scores
    for _, row in user_candidates.iterrows():

        isbn = row["ISBN"]
        book_idx = isbn_to_item_index.get(isbn)

        if book_idx is None or book_idx not in candidate_indices:
            continue

        similarity_score = float(row["Similarity Score"])

        if similarity_sum.get(book_idx, 0) == 0:
            continue

        collaborative_score = (
            weighted_rating_sum[book_idx] / similarity_sum[book_idx]
        )

        # Convert collaborative rating to a 0-1 scale
        normalized_collaborative_score = collaborative_score / 10

        hybrid_score = (
            content_weight * similarity_score
            + collaborative_weight * normalized_collaborative_score
        )

        hybrid_results.append({
            "User-ID": user_id,
            "ISBN": isbn,
            "Content Score": similarity_score,
            "Collaborative Score": collaborative_score,
            "Hybrid Score": hybrid_score
        })

# Create final Hybrid results
hybrid_recommendations_df = pd.DataFrame(hybrid_results)

# Add book details
if not hybrid_recommendations_df.empty:

    hybrid_recommendations_df = hybrid_recommendations_df.merge(
        book_details_lookup,
        on="ISBN",
        how="left"
    )

    hybrid_recommendations_df = hybrid_recommendations_df.sort_values(
        ["User-ID", "Hybrid Score"],
        ascending=[True, False]
    )

    hybrid_recommendations = {
        user_id: group.head(10).reset_index(drop=True)
        for user_id, group in hybrid_recommendations_df.groupby("User-ID")
    }

    print("HYBRID RECOMMENDER RESULTS")
    print("-" * 45)

    print("Total Hybrid recommendations:",
          len(hybrid_recommendations_df))

    print("Users with Hybrid recommendations:",
          len(hybrid_recommendations))

    print("\nSample Hybrid recommendations:")

    sample_hybrid_user = next(iter(hybrid_recommendations))

    print("User ID:", sample_hybrid_user)

    print(
        hybrid_recommendations[sample_hybrid_user][
            ["Book-Title", "Book-Author",
             "Content Score", "Collaborative Score", "Hybrid Score"]
        ].to_string(index=False)
    )

else:
    print("No Hybrid recommendations could be generated.")
```

```python

# Generate expanded Hybrid recommendations

expanded_hybrid_results = []

index_to_isbn = {
    index: isbn for isbn, index in book_mapping.items()
}

content_groups = content_candidates_df.groupby("User-ID")

for user_id, user_candidates in content_groups:

    user_idx = user_mapping.get(user_id)

    if user_idx is None:
        continue

    user_ratings = train_user_book_matrix[user_idx].toarray().flatten()

    rated_indices = set(np.where(user_ratings > 0)[0])

    # Store content scores by ISBN
    content_scores = (
        user_candidates.groupby("ISBN")["Similarity Score"]
        .max()
        .to_dict()
    )

    # Calculate collaborative scores from existing Item-CF neighbours
    weighted_rating_sum = {}
    similarity_sum = {}

    for rated_idx in rated_indices:

        rating = user_ratings[rated_idx]

        neighbour_indices = item_indices_50[rated_idx]
        neighbour_similarities = item_similarities_50[rated_idx]

        for neighbour_idx, similarity in zip(
            neighbour_indices, neighbour_similarities
        ):

            neighbour_idx = int(neighbour_idx)
            similarity = float(similarity)

            if neighbour_idx in rated_indices or similarity <= 0:
                continue

            weighted_rating_sum[neighbour_idx] = (
                weighted_rating_sum.get(neighbour_idx, 0)
                + similarity * rating
            )

            similarity_sum[neighbour_idx] = (
                similarity_sum.get(neighbour_idx, 0)
                + similarity
            )

    collaborative_scores = {}

    for book_idx, weighted_sum in weighted_rating_sum.items():

        isbn = index_to_isbn.get(book_idx)

        if isbn is None or similarity_sum[book_idx] == 0:
            continue

        collaborative_scores[isbn] = (
            weighted_sum / similarity_sum[book_idx]
        ) / 10

    # Combine the candidate lists from both approaches
    all_candidate_isbns = (
        set(content_scores) | set(collaborative_scores)
    )

    for isbn in all_candidate_isbns:

        has_content = isbn in content_scores
        has_collaborative = isbn in collaborative_scores

        content_score = content_scores.get(isbn, np.nan)
        collaborative_score = collaborative_scores.get(isbn, np.nan)

        # Average available signals without penalizing missing signals
        available_scores = []

        if has_content:
            available_scores.append(content_score)

        if has_collaborative:
            available_scores.append(collaborative_score)

        hybrid_score = float(np.mean(available_scores))

        if has_content and has_collaborative:
            signal = "Content + Item-CF"
        elif has_content:
            signal = "Content only"
        else:
            signal = "Item-CF only"

        expanded_hybrid_results.append({
            "User-ID": user_id,
            "ISBN": isbn,
            "Content Score": content_score,
            "Collaborative Score": collaborative_score,
            "Hybrid Score": hybrid_score,
            "Signals": signal
        })

# Create expanded recommendation dataframe
expanded_hybrid_df = pd.DataFrame(expanded_hybrid_results)

if not expanded_hybrid_df.empty:

    expanded_hybrid_df = expanded_hybrid_df.merge(
        book_details_lookup,
        on="ISBN",
        how="left"
    )

    expanded_hybrid_df = expanded_hybrid_df.sort_values(
        ["User-ID", "Hybrid Score"],
        ascending=[True, False]
    )

    expanded_hybrid_recommendations = {
        user_id: group.head(10).reset_index(drop=True)
        for user_id, group in expanded_hybrid_df.groupby("User-ID")
    }

    print("EXPANDED HYBRID RESULTS")
    print("-" * 45)

    print("Total candidates:", len(expanded_hybrid_df))
    print("Users:", len(expanded_hybrid_recommendations))

    print("\nRecommendations per user:")
    print(
        expanded_hybrid_df.groupby("User-ID")
        .size()
        .describe()
    )

    print("\nSignal distribution:")
    print(expanded_hybrid_df["Signals"].value_counts())

    print("\nSample recommendations:")

    sample_user = next(iter(expanded_hybrid_recommendations))

    print(
        expanded_hybrid_recommendations[sample_user][
            ["Book-Title", "Book-Author", "Hybrid Score", "Signals"]
        ].to_string(index=False)
    )

else:
    print("No expanded recommendations generated.")
```

```python
import pandas as pd

# Work with a copy of the complete candidate pool
hybrid_candidates = balanced_hybrid_df.copy()

# Create a normalized title for duplicate checking
hybrid_candidates["title_key"] = (
    hybrid_candidates["Book-Title"]
    .fillna("")
    .str.lower()
    .str.strip()
)

final_hybrid_list = []

for user_id, user_candidates in hybrid_candidates.groupby("User-ID"):

    # Sort all recommendations by hybrid score
    user_candidates = user_candidates.sort_values(
        "Hybrid Score",
        ascending=False
    )

    # Select recommendations that use content-based information
    content_pool = user_candidates[
        user_candidates["Signals"] != "Item-CF only"
    ]

    # Remove duplicate titles and reserve up to 2 slots
    content_selected = (
        content_pool
        .drop_duplicates(subset="title_key")
        .head(2)
    )

    # Find remaining recommendations
    remaining_pool = user_candidates[
        ~user_candidates.index.isin(content_selected.index)
    ]

    # Avoid repeating titles already selected
    remaining_pool = remaining_pool[
        ~remaining_pool["title_key"].isin(
            content_selected["title_key"]
        )
    ]

    # Fill the remaining slots with the highest-scoring candidates
    remaining_selected = (
        remaining_pool
        .drop_duplicates(subset="title_key")
        .head(10 - len(content_selected))
    )

    # Combine both selections
    user_final = pd.concat(
        [content_selected, remaining_selected]
    )

    final_hybrid_list.append(user_final)


# Combine recommendations for all users
final_hybrid_recommendations = pd.concat(
    final_hybrid_list,
    ignore_index=True
)

# Sort recommendations by user and score
final_hybrid_recommendations = (
    final_hybrid_recommendations
    .sort_values(
        ["User-ID", "Hybrid Score"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

# Remove the temporary column
final_hybrid_recommendations = (
    final_hybrid_recommendations.drop(columns=["title_key"])
)


print("FINAL HYBRID RECOMMENDATION RESULTS")
print("---------------------------------------------")

print("Total recommendations:", len(final_hybrid_recommendations))
print("Total users:", final_hybrid_recommendations["User-ID"].nunique())

print("\nSignal distribution:")
print(final_hybrid_recommendations["Signals"].value_counts())

print("\nRecommendations per user:")
print(final_hybrid_recommendations.groupby("User-ID").size().describe())

print("\nDuplicate user-title recommendations:")
print(
    final_hybrid_recommendations.duplicated(
        subset=["User-ID", "Book-Title"]
    ).sum()
)

print("\nSample recommendations:")
display(
    final_hybrid_recommendations[
        ["User-ID", "Book-Title", "Book-Author", "Hybrid Score", "Signals"]
    ].head(20)
)
```
